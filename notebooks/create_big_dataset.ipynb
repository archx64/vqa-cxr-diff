```python
import pandas as pd

df_pair_qa = pd.read_csv('../data/medical-diff-vqa-1.0.1/mimic_pair_questions.csv')
df_pair_qa.shape
```

```python
df_pair_qa.head()
```

```python
df_pair_qa_train = df_pair_qa[df_pair_qa['split'] == 'train']
df_pair_qa_val = df_pair_qa[df_pair_qa['split'] == 'val']
df_pair_qa_test = df_pair_qa[df_pair_qa['split'] == 'test']
df_pair_qa_train.shape, df_pair_qa_val.shape, df_pair_qa_test.shape
```

```python
df_pair_qa_train = df_pair_qa_train.drop(columns=['split'])
df_pair_qa_val = df_pair_qa_val.drop(columns=['split'])
df_pair_qa_test = df_pair_qa_test.drop(columns=['split'])
df_pair_qa_train.shape, df_pair_qa_val.shape, df_pair_qa_test.shape
```

```python
df_all = pd.read_csv('../data/medical-diff-vqa-1.0.1/mimic_all.csv')
df_all.shape
```

```python
df_all.head()
```

```python
df_all['split'].unique()
```

```python
df_all_train = df_all[df_all['split'] == 'train']
df_all_val = df_all[df_all['split'] == 'validate']
df_all_test = df_all[df_all['split'] == 'test']
df_all_train.shape, df_all_val.shape, df_all_test.shape
```

```python
df_all_train = df_all_train.drop(columns=['split'])
df_all_val = df_all_val.drop(columns=['split'])
df_all_test = df_all_test.drop(columns=['split'])
df_all_train.shape, df_all_val.shape, df_all_test.shape
```

```python
df_pair_qa_train.to_csv('E:\\MIMIC-CXR-JPG-224x224\\mimic_pair_questions_train.csv', index=False)
df_pair_qa_val.to_csv('E:\\MIMIC-CXR-JPG-224x224\\mimic_pair_questions_val.csv', index=False)
df_pair_qa_test.to_csv('E:\\MIMIC-CXR-JPG-224x224\\mimic_pair_questions_test.csv', index=False)
df_all_train.to_csv('E:\\MIMIC-CXR-JPG-224x224\\mimic_all_train.csv', index=False)
df_all_val.to_csv('E:\\MIMIC-CXR-JPG-224x224\\mimic_all_val.csv', index=False)
df_all_test.to_csv('E:\\MIMIC-CXR-JPG-224x224\\mimic_all_test.csv', index=False)
```
