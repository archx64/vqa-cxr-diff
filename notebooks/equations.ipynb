## Add Counterfactual Representation Learning

To mitigate spurious correlations and ensure the generative head relies on true clinical differences rather than dataset biases, we introduce a counterfactual learning framework. ## Auxillary Disease Classification 

To explicitly ground the visual features in clinical pathology, we apply an auxiliary multi-label classification head predicting $C=14$ common thoracic diseases. The classification loss is computed for both the reference and current images using Binary Cross-Entropy (BCE):

$$
\mathcal{L}_{\text{BCE}}(y, \hat{y}) = -\frac{1}{C} \sum_{c=1}^{C} \left[ y_c \log(\hat{y}_c) + (1 - y_c) \log(1 - \hat{y}_c) \right]
$$

where $y_c \in \{0, 1\}$ represents the ground truth label for disease $c$, and $\hat{y}_c \in [0, 1]$ represents the predicted probability output by the sigmoid activation function. 

The total auxiliary classification loss applied during the warmup and main training stages is the sum of the losses from both temporal states:

$$
\mathcal{L}_{\text{cls}} = \mathcal{L}_{\text{BCE}}(y_{\text{ref}}, \hat{y}_{\text{ref}}) + \mathcal{L}_{\text{BCE}}(y_{\text{cur}}, \hat{y}_{\text{cur}})
$$
